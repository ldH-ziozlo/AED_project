```python
import requests
import pandas as pd
import xml.etree.ElementTree as ET
```

```python
url = "http://apis.data.go.kr/B552657/AEDInfoInqireService/getEgytAedManageInfoInqire?"

service_key = "clO+/PLnuHzh4HnDTskKmUoNxNSV3WiNk7W+fHaAEVgL7qvGErwtrfIr208DTWvrkDdJHTYgQpkD63hOf4zHxw=="
```

```python
params = {
    "serviceKey": service_key,
    "pageNo": 1,
    "numOfRows": 10
}

response = requests.get(url, params=params)

print(response.status_code)
print(response.text[:1000])
```

```python
all_rows = []

page = 1
num_rows = 1000

while True:
    
    params = {
        "serviceKey": service_key,
        "pageNo": page,
        "numOfRows": num_rows
    }
    
    response = requests.get(url, params=params)
    
    if response.status_code != 200:
        print("요청 오류:", response.status_code)
        break
    
    root = ET.fromstring(response.text)
    
    result_code = root.findtext(".//resultCode")
    result_msg = root.findtext(".//resultMsg")
    
    if result_code != "00":
        print("API 오류:", result_code, result_msg)
        break
    
    items = root.findall(".//item")
    
    if len(items) == 0:
        break
    
    for item in items:
        row = {}
        
        for child in item:
            row[child.tag] = child.text
        
        all_rows.append(row)
    
    total_count = int(root.findtext(".//totalCount"))
    
    print(
        f"{page}페이지 완료 | "
        f"현재 {len(all_rows):,}건 / 전체 {total_count:,}건"
    )
    
    if len(all_rows) >= total_count:
        break
    
    page += 1
```

```python
aed = pd.DataFrame(all_rows)

print(aed.shape)

aed.head()
```

```python
aed["wgs84Lat"] = pd.to_numeric(
    aed["wgs84Lat"],
    errors="coerce"
)

aed["wgs84Lon"] = pd.to_numeric(
    aed["wgs84Lon"],
    errors="coerce"
)
```

```python
aed_location_count = (
    aed
    .dropna(subset=["wgs84Lat", "wgs84Lon"])
    [["wgs84Lat", "wgs84Lon"]]
    .drop_duplicates()
    .shape[0]
)

print("AED 전체 대수:", len(aed))
print("AED 고유 위치 수:", aed_location_count)
```
